## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Bagaimana hubungan kondisi cuaca dengan jumlah penyewaan sepeda harian pada sistem Bike Sharing selama periode 2011–2012?

Keterangan:

- **Specific**: Fokus pada faktor yang memengaruhi penurunan jumlah penyewaan sepeda harian berdasarkan kondisi cuaca, seperti temperatur, kelembapan, kecepatan angin, dan situasi cuaca. Bukan sekadar melihat perubahan jumlah penyewaan secara umum.
- **Measurable**: menggunakan variabel `cnt` sebagai jumlah penyewaan dan variabel cuaca seperti `temp`, `hum`, `windspeed`, dan `weathersit`.
- **Action-Oriented**: Dengan mengetahui faktor penyebab penurunan penyewaan, pengelola dapat menentukan strategi seperti penyesuaian jumlah armada atau membuat program promosi pada kondisi cuaca tertentu.
- **Relevant**: Penurunan jumlah penyewaan sepeda berkaitan langsung dengan faktor lingkungan karena kondisi cuaca dapat memengaruhi perilaku pengguna dalam menggunakan layanan bike sharing.
- **Time-bound**: Dibatasi pada periode tahun 2011–2012 sesuai dengan data historis penyewaan sepeda yang tersedia.

- **Pertanyaan 2:** Bagaimana perbedaan pola jumlah penyewaan sepeda antara hari kerja dan hari libur selama periode Januari 2011 hingga Desember 2012?

Keterangan:

- **Specific**: Fokus pada perbandingan pola penggunaan sepeda antara hari kerja dan hari libur berdasarkan jumlah penyewaan. Bukan hanya menganalisis jumlah pengguna secara keseluruhan.
- **Measurable**: Ada data yang dapat dihitung menggunakan jumlah penyewaan sepeda (cnt), status hari kerja (workingday), dan status hari libur (holiday) untuk melihat perbedaannya.
- **Action-Oriented**: Hasil analisis dapat digunakan untuk menentukan strategi operasional, seperti pengaturan ketersediaan sepeda pada hari dengan tingkat permintaan tinggi.
- **Relevant**: Analisis ini relevan karena membantu memahami kebiasaan pengguna sehingga layanan bike sharing dapat disesuaikan dengan kebutuhan pengguna.
- **Time-bound**: Analisis dibatasi pada periode Januari 2011 hingga Desember 2012 agar hasil memiliki batas waktu yang jelas.

### Pertanyaan 2: Bagaimana perbedaan pola jumlah penyewaan sepeda antara hari kerja dan hari libur selama periode Januari 2011 hingga Desember 2012?

```python
hour_analysis = hour_df.copy()


hour_analysis['workingday_label'] = (
    hour_analysis['workingday']
    .replace({
        1: 'Hari Kerja',
        0: 'Hari Libur'
    })
)


hour_analysis.head()
```

```python
hour_pattern = (
    hour_analysis
    .groupby(
        [
            'hr',
            'workingday_label'
        ]
    )['cnt']
    .mean()
    .reset_index()
)


hour_pattern.head()
```

```python
plt.figure(figsize=(12,6))


sns.lineplot(
    data=hour_pattern,
    x='hr',
    y='cnt',
    hue='workingday_label',
    marker='o'
)


plt.title(
    'Pola Rata-rata Penyewaan Sepeda Berdasarkan Jam\nHari Kerja vs Hari Libur'
)


plt.xlabel(
    'Jam (0-23)'
)


plt.ylabel(
    'Rata-rata Jumlah Penyewaan'
)


plt.xticks(range(0,24))

plt.grid(True)

plt.show()
```

```python
peak_workingday = (
    hour_analysis[
        hour_analysis['workingday_label']=='Hari Kerja'
    ]
    .groupby('hr')['cnt']
    .mean()
    .sort_values(ascending=False)
    .head(3)
)


peak_holiday = (
    hour_analysis[
        hour_analysis['workingday_label']=='Hari Libur'
    ]
    .groupby('hr')['cnt']
    .mean()
    .sort_values(ascending=False)
    .head(3)
)


print("Peak Hari Kerja")
display(peak_workingday)


print("Peak Hari Libur")
display(peak_holiday)
```

## Conclusion & Recommendation

- **Conclusion pertanyaan 1:** Berdasarkan hasil analisis pengaruh kondisi cuaca terhadap jumlah penyewaan sepeda, ditemukan adanya perbedaan rata-rata jumlah penyewaan berdasarkan kondisi cuaca. Pada kondisi cuaca cerah (weathersit 1), rata-rata jumlah penyewaan mencapai 4.876,79 sepeda, lebih tinggi dibandingkan kondisi berawan (weathersit 2) dengan rata-rata 4.035,86 sepeda. Sementara itu, kondisi cuaca kurang mendukung seperti hujan ringan atau salju ringan (weathersit 3) memiliki rata-rata penyewaan yang lebih rendah yaitu 1.803,29 sepeda. Hasil analisis menunjukkan bahwa kondisi cuaca yang lebih baik berkaitan dengan peningkatan rata-rata jumlah penyewaan sepeda, sedangkan kondisi cuaca yang kurang mendukung berkaitan dengan penurunan aktivitas penyewaan. Berdasarkan temuan tersebut, operator Bike Sharing dapat mempertimbangkan faktor cuaca dalam melakukan perencanaan operasional, seperti pengaturan ketersediaan armada dan strategi pelayanan pada periode dengan potensi permintaan tinggi.

- **Conclusion pertanyaan 2:** Berdasarkan hasil analisis pola penyewaan sepeda berdasarkan jam penggunaan, terdapat perbedaan karakteristik penggunaan antara hari kerja dan hari libur. Pada hari kerja, pola penyewaan menunjukkan dua periode peningkatan utama, yaitu pada pagi hari sekitar pukul 08.00 dan sore hari sekitar pukul 17.00–18.00. Pola tersebut menunjukkan bahwa penggunaan sepeda pada hari kerja memiliki karakteristik yang berkaitan dengan aktivitas mobilitas rutin, seperti perjalanan menuju kantor atau sekolah serta perjalanan pulang. Sementara itu, pada hari libur, pola penyewaan cenderung lebih stabil dengan peningkatan penggunaan pada periode siang hari sekitar pukul 10.00–15.00, yang mengindikasikan adanya aktivitas rekreasi atau kegiatan santai. Hasil analisis ini menunjukkan bahwa faktor waktu penggunaan menjadi salah satu aspek penting dalam pengelolaan layanan Bike Sharing karena kebutuhan pengguna memiliki pola yang berbeda antara hari kerja dan hari libur.

**Rekomendasi Action Item:**
- Operator Bike Sharing dapat menggunakan informasi kondisi cuaca sebagai dasar dalam pengelolaan armada. Pada kondisi cuaca cerah dengan rata-rata penyewaan tertinggi sebesar 4.876,79 sepeda, ketersediaan armada perlu dipastikan optimal untuk memenuhi permintaan pengguna. Sebaliknya, ketika kondisi cuaca kurang mendukung dengan rata-rata penyewaan sebesar 1.803,29 sepeda, operator dapat melakukan penyesuaian jumlah armada yang tersedia serta mengoptimalkan kegiatan operasional lainnya.
- Operator Bike Sharing dapat melakukan redistribusi armada berdasarkan pola waktu penggunaan pengguna. Pada hari kerja, penyediaan sepeda perlu difokuskan pada jam sibuk pagi dan sore yang menunjukkan pola mobilitas komuter, sedangkan pada hari libur distribusi armada dapat diarahkan ke lokasi dengan aktivitas rekreasi tinggi pada periode siang hari. Strategi ini dapat membantu meningkatkan efisiensi penggunaan armada serta menjaga ketersediaan sepeda sesuai kebutuhan pengguna.
